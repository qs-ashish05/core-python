```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.arange(-20, 20)
data
```

### Sigmoid Activation Function

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

result = []
for i in data:
    result.append(sigmoid(i))

result
```

```python
plt.plot(data, result)
plt.title("Sigmoid function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## Tanh Activation Function

```python
def tanh(x):
    return (np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x))


result = []
for i in data:
    result.append(tanh(i))
result
```

```python
plt.plot(data, result)
plt.title("Tanh function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## ReLu Activation Function

```python
def relu(x):
    if x <= 0:
        return 0
    else:
        return x


result = []
for i in data:
    result.append(relu(i))

result
```

```python
plt.plot(data, result)
plt.title("ReLu function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## LeakyReLu Activation Function

```python
def Leakly_ReLu(x):
    if x <= 0:
        return 0.1*x
    else:
        return x

result = []
for i in data:
    result.append(Leakly_ReLu(i))

result
```

```python
plt.plot(data, result)
plt.title("Leakly ReLu function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## ELU Activation Function

```python
def ELU(x):
    a = 0.1
    if x > 0:
        return x
    else:
        return a*(np.exp(x) - 1)

result = []
for i in data:
    result.append(ELU(i))

```

```python
plt.plot(data, result)
plt.title("ELU function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## Softmax Activation Function

```python
sum = 0.0
for i in data:
    sum = sum + np.exp(i)

def softmax(x):
    return np.exp(i) / sum

result = []
for i in data:
    result.append(softmax(i))

result
```

```python
plt.plot(data, result)
plt.title("softmax function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## Softplus Activation Function

```python
def softplus(x):
    return np.log(1 + np.exp(x))

result = []
for i in data:
    result.append(softplus(i))

result
```

```python
plt.plot(data, result)
plt.title("softplus function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```

## Binary Step Activation Function

```python
def binary_step(x):
    return 1 if x>0 else 0

result = []
for i in data:
    result.append(binary_step(i))
```

```python
plt.plot(data, result)
plt.title("binary function")
plt.xlabel("data")
plt.ylabel("result")
plt.show()
```
