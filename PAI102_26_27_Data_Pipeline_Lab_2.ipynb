## Data Pipeline Using CSV, JSON, Avro, and Parquet

---


### Objective
To **read and validate all generated file formats**


## Task 1: Load Required Libraries

```python

import pandas as pd
import json
from fastavro import reader
import pyarrow.parquet as pq
import pyarrow as pa
```

## Task 2: Read CSV File

```python

df_csv = pd.read_csv("students.csv")
df_csv
```

## Task 3: Read JSON File

```python

with open("students.json", "r") as f:
    json_data = json.load(f)

df_json = pd.DataFrame(json_data)
df_json
```

## Task 4: Read Avro File

```python

avro_records = []
with open("students.avro", "rb") as f:
    for record in reader(f):
        avro_records.append(record)

df_avro = pd.DataFrame(avro_records)
df_avro
```

## Task 5: Read Parquet File

```python

df_parquet = pq.read_table("students.parquet").to_pandas()
df_parquet
```

## Task 6: Data Consistency Check

```python

print("CSV rows:", len(df_csv))
print("JSON rows:", len(df_json))
print("Avro rows:", len(df_avro))
print("Parquet rows:", len(df_parquet))
```
