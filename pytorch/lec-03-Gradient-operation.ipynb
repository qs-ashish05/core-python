## Gradient in neural network

```python
x = torch.tensor(6.7)
y = torch.tensor(0.0)
```

```python
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
```

```python
z = w*x + b
z
```

```python
y_pred = torch.sigmoid(z)
y_pred
```

```python
def binary_cross_loss(y_pred, y):
    e = 1e-8
    prediction = torch.clamp(y_pred, e, 1-e)
    return -(y*torch.log(prediction) + (1-y)*torch.log(1-prediction))
```

```python
loss = binary_cross_loss(y_pred, y)
loss
```

```python
loss.backward()
```

```python
w.grad  # dloss/dw
```

```python
b.grad  # dloss / db
```
