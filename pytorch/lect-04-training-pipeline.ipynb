```python
# numpy arrays to pytorch tensors

x_train_tensor = torch.from_numpy(x_train)
x_test_tesnor = torch.from_numpy(x_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

## Defining the model

```python
class mySimpleNeuralNetwork:

    # here x is basically x_train_tensor
    def __init__(self, x):
        self.weights = torch.rand(x.shape[1], 1, dtype=torch.float64, requires_grad=True)
        self.biases = torch.zeros(1, dtype=torch.float64, requires_grad=True)

    def forward(self, x):
        # z = w*x + b
        z = torch.matmul(x, self.weights) + self.biases

        # y_pred = sigmoid(z)
        y_pred = torch.sigmoid(z)

        return y_pred

    def loss_functon(self, y_pred, y):
        e = 1e-7
        prediction = torch.clamp(y_pred, e, 1-e)
        loss =  -(y*torch.log(prediction) + (1-y)*torch.log(1-prediction)).mean()

        return loss
```

```python
learning_rate = 0.1
epochs = 25
```

### Training Pipeline

```python
model = mySimpleNeuralNetwork(x_train_tensor)
# model.weights
# in loop:
    # forward pass
    # Calculate loss
    # backward pass - gradient calculation
    # parameters
```

```python
for epoch in range(epochs):
    # 1. Forward pass
    y_pred = model.forward(x_train_tensor)

    # 2. Calculate loss
    loss = model.loss_functon(y_pred, y_train_tensor)

    # 3. Backward pass (calculates gradients)
    loss.backward()

    # 4. Update parameters and reset gradients
    with torch.no_grad():
        # In-place parameter updates (sub_ modifies values without breaking the graph)
        model.weights.sub_(learning_rate * model.weights.grad)
        model.biases.sub_(learning_rate * model.biases.grad)

        # Zero out gradients for the next iteration (PyTorch accumulates them by default)
        model.weights.grad.zero_()
        model.biases.grad.zero_()

    print(f"Epoch {epoch+1}, Loss: {loss.item()}")
```

```python
with torch.no_grad():
    y_pred = model.forward(x_test_tesnor)
```
