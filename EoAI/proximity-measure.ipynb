## Nominal Attributes - dissimilarity Measure

    - d(i,j) = (p -m)/p
    - p is the number of attributes in dataframe (other than the identifiier column )
    - m is the number of similar points for given pair

```python
import pandas as pd

data = {
    "Roll No": [1,2,3,4,5],
    "Grade": ['A','B','C','A','E']
}


# convert the data into pandas dataframe
df = pd.DataFrame(data).set_index("Roll No")
```

```python
df.head()
```

```python
def dissimilarity_mtx_for_nominal(df, p):

    n = len(df)

    mtx = [[0.0 for _ in range(n)] for _ in range(n)]

    for i in range(n):
        for j in range(i+1, n):

            m = 0  # count of similar points
            if(df.iloc[i]['Grade'] == df.iloc[j]['Grade']):
                m += 1
            dis = (p-m) / p
            print(f'Calulating {i} and {j} dissimilarity score: {dis}')

            # mtx[i][j] = dis # for upper triangular mtx
            mtx[j][i] = dis
    return mtx
```

```python
mtx = dissimilarity_mtx_for_nominal(df, 1)
mtx
```

### Below code is more gneric formula this will work for all type of dataframes

```python
def nominal_dissimilarity_measure(df):
    n = len(df)
    p = df.shape[1]   # number of attributes

    mtx = [[0.0 for _ in range(n)] for _ in range(n)]

    for i in range(n):
        for j in range(i + 1, n):

            m = 0

            # Compare every attribute
            for column in df.columns:
                if df.iloc[i][column] == df.iloc[j][column]:
                    m += 1

            dis = (p - m) / p

            # Matrix is symmetric
            # mtx[i][j] = dis
            mtx[j][i] = dis

    return mtx
```

```python
mtx = nominal_dissimilarity_measure(df1)
mtx
```
