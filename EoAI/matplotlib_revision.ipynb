```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4], )
plt.ylabel('some numbers')
plt.show()
```

```python
plt.plot([1, 2, 3, 4], marker='o', color='red')
plt.ylabel('some numbers')
plt.show()
```

```python
plt.plot(
    [1, 2, 3, 4],
    color='blue',          # line color
    marker='o',
    markerfacecolor='red', # dot color
    markeredgecolor='yellow'
)

plt.ylabel('some numbers')
plt.show()
```
