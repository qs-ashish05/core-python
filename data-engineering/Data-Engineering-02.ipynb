```python
# Avro
import fastavro
```

```python
import  json

schema = {
    "type": "record",
    "name": "Student",
    "fields": [
        {"name": "student_id", "type": "int"},
        {"name": "name", "type": "string"},
        {"name": "department", "type": "string"},
        {"name": "marks", "type": "float"},
        {"name": "ispassed", "type": "boolean"},
    ]
}
```

```python
parsed_schema = fastavro.parse_schema(schema)

with open('sample_data2.json') as f:
    records = json.load(f)
```

```python
records
```

```python

with open('students.avro', 'wb') as f:
    fastavro.writer(f, parsed_schema, records)
```

```python
import pyarrow as pq
```

```python
with open('students.avro', 'rb') as f:
    records = list(fastavro.reader(f))
```

```python
table = pq.Table.from_pylist(records)
```

```python
table
```

```python
pq.parquet.write_table(table, 'example.parquet')
```
