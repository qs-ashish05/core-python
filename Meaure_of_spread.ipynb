```python
df = pd.read_csv("./gapminder_full.csv")
```

```python
c = 1.5
d = 5
```

```python
df.head(5)
```

```python
df['transformed_value'] = c*df['life_exp'] + d
df.head(5)
```

```python
# new variables

mean_t = df['transformed_value'].mean()
print("Transformed data Mean:", mean_t)

median_t = df['transformed_value'].median()
print("Transformed data Median:", median_t)

mode_t = df['transformed_value'].mode()
print("Transformed data Mode:",mode_t)

range_t = df['transformed_value'].max() - df['life_exp'].min()
print("Transformed data Range:", round(range_t,4))

var_t = df['transformed_value'].var()
print("Transformed data Variance:", var_t)

std_t = df['transformed_value'].std()
print("Transfomed data Standard Deviation:", std_t)
```

```python
# Statistics
metrics = [
    "Mean",
    "Median",
    "Mode",
    "Range",
    "Variance",
    "Std Deviation"
]

old_values = [
    mean,
    median,
    mode.iloc[0],       # first mode
    range,
    var,
    std
]

new_values = [
    mean_t,
    median_t,
    mode_t.iloc[0],      # first mode
    range_t,
    var_t,
    std_t
]

# Position of bars
x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, old_values, width, label="Old Values")
plt.bar(x + width/2, new_values, width, label="new values")

plt.xticks(x, metrics)
plt.ylabel("Value")
plt.title("Comparison of Statistical Measures")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
change = ((mean_t - mean) / mean) * 100
print(f"Change in mean: {change:.2f}%")
```
