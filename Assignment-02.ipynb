1. roll dice until we get all three dice with 6 and print the number of trials

```python
import random

count = 0
while True:
    a = random.randint(1,6)
    b = random.randint(1,6)
    c = random.randint(1,6)

    count += 1
    if (a == 6 and b == 6 and c == 6):
        print("All 3 numbers are six in count ", count)
        break
```

```python
count = 0
rolls = 0
while (count != 3):
    a = random.randint(1,6)
    if(a == 6):
        count += 1
    else:
        count = 0
    rolls += 1

print(rolls)
```

## 2. 1 to 50 print according to multiple of
    - 3 --> fizz
    - 5 --> buzz
    - 7 --> bang

```python
for i in range(1, 51):
    isMultiple = False
    result = ""
    if(i%3 == 0):
        result += "Fizz"
        isMultiple = True
    if(i%5 == 0):
        result += "Buzz"
        isMultiple = True
    if(i%7 == 0):
        result += "Bang"
        isMultiple = True

    if(isMultiple):
        print(f"{i} - {result}")
```

## 3. Maximum attempt of PIN

```python
correct_pin = 7856
attempts = 3
pin = ""
while (pin != correct_pin):
    attempts -= 1
    pin = input("Enter your PIN: ")
    if (pin == correct_pin):
        print("Corerct PIN")
        break
    if (pin != correct_pin):
        print("Wrong PIN try again")
    if (attempts == 0):
        print("Maximum attempts reached - card block")
        break
```

## 5. Longest streak of temprature

```python

temperatures = [32, 36, 38, 40, 37, 34, 39, 41, 42, 43, 36, 33, 35, 37, 39]

longest = 0
current = 0
start_index = 0
current_start = 0

for i in range(len(temperatures)):
    if (temperatures[i] >= 35):
        if (current == 0):
            current_start = i

        current += 1

        if (current > longest):
            longest = current
            start_index = current_start

    else:
        current = 0

print("Length:", longest)
print("Start index:", start_index)
```

## 6. Secret Santa Shuffler

```python
names = ['A', 'B', 'C', 'D', 'E', 'F']

paired = [False] * len(names)
pairs = []

for i in range(len(names)):


    if paired[i]:
        continue

    while True:

        a = random.randint(0, len(names) - 1)
        if paired[a] == False and a != i:
            paired[i] = True
            paired[a] = True

            pairs.append([names[i], names[a]])
            break

print(pairs)
```
